```python
import requests
import bs4
import pandas as pd
```

```python
url = "https://realpython.github.io/fake-jobs/"
response = requests.get(
    url,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=15,
)
response.raise_for_status()

soup = bs4.BeautifulSoup(response.text, "html.parser")
```

# creating dataframe

importing pandas for simple data handling

```python
job = pd.DataFrame()
```

```python
job["job title"] = [str(x)[23:-5] for x in soup.select("h2")]
job["company name"] = [str(x)[34:-5] for x in soup.select("h3")]
job["location"] =  [str(x)[20:-5] for x in soup.select(".location")]
job["Job detail page URL"] = [link.get("href") for idx,link in enumerate(soup.select('a')) if idx%2 != 0]
```

```python
job
```

converting into csv file

```python
job.to_csv("job.csv",index = False)
```
